## Paired associate learning from instructions model and task

The model file for this task is [paired-learning-model.lisp](../unit7/paired-learning-model.lisp).

```python
paired.experiment(1)
```

## Production Compilation issues model in the modeling text

The starting model for this task is [production-compilation-issues-model.lisp](../unit7/production-compilation-issues-model.lisp).

Run this cell to connect to ACT-R and load the [production compilation issues task](pcomp_issues.py) and model.  


```python
import actr
import pcomp_issues
```

Run this cell to run the model on one trial.  


```python
pcomp_issues.trials(1)
```

Run this cell to run the model through the whole task (150 trials) the specified number of times and report the average results.  
The model trace should be turned off before running this to avoid the large and slow trace output.

```python
pcomp_issues.game(10)
```

Run this cell to run the model after making the change to the retrieved-a-non-win production as described in the text to find a game where that production is used.  


```python
used = False
def set_flag ():
    global used
    used = True
actr.add_command('set-flag',set_flag,'Command for testing the pcomp-issues retrieved-a-non-win production.')
while used == False:
    pcomp_issues.game(1)
```

Run this cell to find productions which have never been matched by looking for a nil (None) :utility parameter value.  


```python
for i in actr.no_output('spp',':name',':utility'):
    if i[1] == None:
        print(i[0])
```

Run this cell to run the model through the whole task the specified number of times and report the result of every game as well as the average results.  
The model trace should be turned off before running this to avoid the large and slow trace output.

```python
pcomp_issues.game(10,True)
```
