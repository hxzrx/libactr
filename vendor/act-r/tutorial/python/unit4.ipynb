## Paired associate model and task

The model file for this task is [paired-model.lisp](../unit4/paired-model.lisp).

Run this cell to connect to ACT-R and load the [paired task](paired.py) and model.  


```python
import actr
import paired
```

Run this cell to run yourself through 2 blocks of the task with 2 items per block.  


```python
paired.task(2,2,True)
```

Run this cell to have the model go through the provided number of pairs of items and trials (2 pairs and 3 trials with the initial values 2,3).  


```python
paired.task(2,3)
```

Run this cell to have the model perform the whole experiment the indicated number of times.  
You should turn off the trace in the model before running the whole experiment to avoid a long delay and excesive amount of trace information.

```python
paired.experiment(2)
```

## Zbrodoff assignment model and task

The model file for this task is [zbrodoff-model.lisp](../unit4/zbrodoff-model.lisp).  


Run this cell to connect to ACT-R and load the [zbrodoff task](zbrodoff.py) and model.  


```python
import actr
import zbrodoff
```

Run this cell to have the model perform one trial of the task with the indicated stimuli.  
The first three values are strings which indicate the problem to display.  
The fourth value is either 'k' or 'd' which represents the correct key press for the trial (k indicating the problem is true and d indicating it is false).  
The fifth parameter is optional, but if specified as True means that a visible window will be displayed so you can see the model's visual attention ring as it does the task.  
Thus, values of: ('a','2','c','k',True) will present the problem a + 2 = c, which is correct so the model should press the k key, and it will be displayed in a visible window.  


```python
zbrodoff.problem('a','2','c','k',True)
```

Run this cell to have the model perform one set of items (24 trials which is one of each possible problem in a random order).  


```python
zbrodoff.set()
```

Run this cell to run the model through one block of items (192 trials which is 8 sequential sets).  


```python
zbrodoff.block()
```

The problem, set, and block functions do not reset the model before running it, which means it will retain the chunks it has learned.  
If you want to reset the model you can run this cell or press the "Reset" button in the Environment GUI.  


```python
actr.reset()
```

Run this cell to run the model once through the whole experiment (3 sequential blocks).  


```python
zbrodoff.experiment()
```

Run this cell to run the model through the experiment the specified number of times reporting the averaged data and their fit to the experimental results.  
Here are the results from a good solution to this task:  
CORRELATION: 0.993  
MEAN DEVIATION: 0.071  
|  |2 (64)|3 (64)| 4 (64)|
|---|---|---|---|
|Block 1| 1.977 (64)| 2.374 (64)| 2.805 (64)|
|Block 2| 1.262 (64)| 1.482 (64)| 1.505 (64)|
|Block 3| 1.095 (64)| 1.154 (64)| 1.219 (64)|
  
An important aspect of that is that the numbers in parentheses, which are the number it got correct, are all the same indicating that the model responded correctly every time.  


```python
zbrodoff.compare(5)
```
