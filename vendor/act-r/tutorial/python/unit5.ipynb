## Fan Effect models and tasks

### Fan Effect model and task with perceptual and motor actions

The model file for this task is [fan-model.lisp](../unit5/fan-model.lisp).

Run this cell to connect to ACT-R and load the [fan task](fan.py) and model which uses perceptual and motor actions.

```python
import actr
import fan
```

Run this cell to have the model perform one trial.  The parameters are strings of the person and location that are presented, whether or not that is a studied item, and either the string 'person' or 'location' to indicate which retrieval production to use (as described in the unit text).  
The parameters in the starting evaluation below ('lawyer','store',True,'person') are asking the model to decide if 'the lawyer is in the store' was one of the studied sentences, the True value indicates that it should respond yes because it was in the set, and it should use the person as the cue to retrieve a chunk to make the decision.  


```python
fan.sentence('lawyer','store',True,'person')
```

Run this cell to have the model perform all trials of the task and report the results.  
You should turn off the trace in the model before running the whole experiment to avoid a long delay and an excesive amount of trace information.  


```python
fan.experiment()
```

### Fan Effect model and task which does not use perceptual and motor actions

The model file for this task is [fan-no-pm-model.lisp](../unit5/fan-no-pm-model.lisp).

Run this cell to connect to ACT-R and load the [fan task](fan_no_pm.py) and model which does not use perceptual and motor actions.  


```python
import actr
import fan_no_pm
```

Run this cell to have the model perform one trial of this version of the task.  
The parameters are similar to the sentence function described above, except that the person and place need to be specified as a string inside of a string, "'lawyer'" instead of "lawyer", because they are being used to set a chunk's slot value to a string (as described in the [unit5 code document](../unit5/unit5_code.pdf) ).  


```python
fan_no_pm.sentence("'lawyer'","'store'",True,'person')
```

Run this cell to have the model perform all trials of the task and report the results.  
You should turn off the trace in the model before running the whole experiment to avoid a long delay and an excesive amount of trace information.  


```python
fan_no_pm.experiment()
```

## 1-hit Blackjack assignment

The model file for this task is [1hit-blackjack-model.lisp](../unit5/1hit-blackjack-model.lisp).  


Run this cell to connect to ACT-R and load the [onehit task](onehit.py) and model.  


```python
import actr
import onehit
```

Run this cell to have the model play the indicated number of hands (without being reset).  


```python
onehit.hands(4)
```

Run this cell to show the model's learning over the specified number of 100 hand games.

```python
onehit.learning(50)
```

If you want the model to run with a different version of the game you can specify the function that implements that game as the third parameter (the second is either True or False to indicate whether to draw the graph of the results).  
Details on creating new games are in the unit5 code document, and an additional game called game1 is included in the task file.  


```python
onehit.learning(50,True,onehit.game1)
```

Note that if you edit the onehit.py file and want to run the model with your new game you will need to either restart the kernel and then import your file, or run this cell to have it explicitly import the onehit module again.  


```python
import importlib
importlib.reload(onehit)
```

## Declarative Memory Related Issues

The model file for the discussion in the modeling text is [declarative-issues.lisp](../unit5/declarative-issues.lisp).  
There is not task that accompanies this model.  
Run this cell to load the model.

```python
actr.load_act_r_model('ACT-R:tutorial;unit5;declarative-issues.lisp')
```

Run this cell to run the model (adjust the time as needed to match the text).  


```python
actr.run(2)
```

Run this cell to call the ACT-R whynot-dm command to get information about the indicated chunks during the most recent retrieval request.  
Replace &lt;chunk-names&gt; with string parameters naming chunks, for example, 'v1','v2','v3'.

```python
actr.whynot_dm(<chunk-names>)
```

Run this cell to change the value of the :act parameter to medium.  


```python
actr.set_parameter_value(':act','medium')
```

Run this cell to call the ACT-R sdp command to see the parameters for chunks in declarative memory.  
Replace &lt;chunk-names&gt; with string parameters naming chunks, for example, 'v1','v2','v3'.

```python
actr.sdp(<chunk-names>)
```

Run this cell to change the value of the :rt parameter to -10.  


```python
actr.set_parameter_value(':rt',-10)
```

Run this cell to call the simulate-retrieval-request command with a description of a retrieval request (in this case the example in the text).  


```python
actr.simulate_retrieval_request('-','result',None,'simple-value-chunk',True)
```

Run this cell to call the ACT-R mp-show-queue command to see the future actions which are scheduled for the model to perform.  


```python
actr.mp_show_queue()
```

Run this cell to call the ACT-R print-dm-finsts command to see the chunks in declarative memory marked with a finst.  


```python
actr.print_dm_finsts()
```

Run this cell to enable the recording of the details of retrieval requests.  


```python
actr.record_history('retrieval-history')
```

Run this cell to get the raw JSON formatted details of recorded retrieval requests.  


```python
actr.get_history_data('retrieval-history')
```

Run this cell to get a list of times and chunks for which history information is available.  


```python
actr.saved_activation_history()
```

Run this cell to get the activation trace from the history information at a particular time in milliseconds (in this case 26591ms).  


```python
actr.print_activation_trace(26591)
```

Run this cell to get the activation trace from the history information for a specific chunk at a particular time in milliseconds (in this case chunk f3 at time 26591ms).  


```python
actr.print_chunk_activation_trace('f3',26591)
```

Run this cell to use the sdm command to search declarative memory for a chunk given a description of slots and values.  


```python
actr.sdm('-','goal','nil')
```
