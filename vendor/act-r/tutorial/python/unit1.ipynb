# Run the ACT-R tutorial unit 1 models

Run the ACT-R software if it is not already running.

Run this cell to connect to ACT-R.


```python
import actr
```

The unit 1 models should be run using the ACT-R Environment as described in the unit.
When you do that, the model trace will be shown in the cell above that shows the connection to ACT-R.
However, you can also run them using the cells below using functions which are described in the unit 2 code document.
If you do that, then the model trace will show in the cell with the run function.

Run this cell to load the [addition model](../unit1/addition.lisp).


```python
actr.load_act_r_model('ACT-R:tutorial;unit1;addition.lisp')
```

Run this cell to load the [semantic model](../unit1/semantic.lisp).


```python
actr.load_act_r_model('ACT-R:tutorial;unit1;semantic.lisp')
```

## This section is for fixing the broken model as described in the unit1 modeling text.


Run this cell to load the [broken addition model](../unit1/broken-addition.lisp).
You will need to do this every time you change and save the model (or you can press the "Reload" button on the Control Panel).


```python
actr.load_act_r_model('ACT-R:tutorial;unit1;broken-addition.lisp')
```

Run this cell to run the currently loaded model.  The time to run can be changed from 10 seconds as needed.


```python
actr.run(10)
```

Replace &lt;name&gt; with the name of the production you want the information on e.g. "initialize-addition".
Then run this cell to call the whynot function (or press the "Why Not?" button in the Procedural tool window of the Environment).


```python
actr.whynot(<name>)
```

Replace &lt;buffers&gt; with the name or comma separated names of the buffers you want the information on e.g. "goal" or "goal","retrieval".
Then run this cell to call the buffer_chunk function (or use the Buffers window of the Environment).


```python
actr.buffer_chunk(<buffers>)
```
