## Building Sticks Task

The model file for this task is [bst-model.lisp](../unit6/bst-model.lisp).  


Run this cell to connect to ACT-R and load the [bst task](bst.py).  


```python
import actr
import bst
```

Run this cell to go through two trials of the task yourself.  


```python
bst.test(1,True)
```

Run this cell to have the model perform two trials of the task which you can watch in the Experiment Window.

```python
bst.test(1,False)
```

Run this cell to have the model perform the whole experiment the indicated number of times and report the average results.

```python
bst.experiment(25)
```

## Choice assignment

The starting model file for this task is [choice-model.lisp](../unit6/choice-model.lisp).  
The start of the task code is in the file [choice.py](../python/choice.py).  


Once you have implemented the model and task, you can run this cell to connect to ACT-R and load the task and model.

```python
import actr
import choice
```

When you change the model you will need to either run this cell to reload the model or press the Reload button in the ACT-R Environment GUI to have the updated model loaded.

```python
actr.reload()
```

When you change the choice.py file you will need to run this cell to reload the choice module.

```python
import importlib
importlib.reload(choice)
```

Run this cell to run your model through one trial of the task with the model function you have written.

```python
choice.model()
```

Run this cell to run your data function which should run your model through the experiment the indicated number of times and report the average results and fit to the data.

```python
choice.data(10)
```
